A class is like a blueprint that defines the attributes and methods something will have. An object is an actual instance created from that class. For example, Student is the class, while student1 is an object created from the Student class.

```python
class Student:
    def __init__(self , name , major , score):
        self.name = name
        self.major = major
        self.score = score
    def display_info(self):
        print("name:", self.name)
        print("major:", self.major)
        print("score:" , self.score)

    def get_letter_grade(self):
      if self.score >= 90:
       return "A"
      elif self.score >= 80:
         return "B"
      elif self.score >= 70:
         return "c"
      elif self.score >= 60:
         return "D"
      else:
         return "F"


    def is_passing(self):
          if self.score >= 60:
             return True
          else:
             return False

                     
student1  = Student("Alice Johnson" , "Computer Science" , 92)
student2 = Student("John mavi" , "cybersecuriy" , 70 )
student3 = Student("Tanner Baucknight" , "Marketing" , 50 )  

Students = [student1 , student2 , student3]

for student in Students:
   if student.is_passing():
      status = "passing"
   else:
      status = "not passing"

   print(student.name, ":", student.get_letter_grade(), "-", status)
# print(student1.get_letter_grade())
# print(student2.get_letter_grade())
# print(student3.get_letter_grade())
# print(student1.is_passing())
# print(student2.is_passing())
# print(student3.is_passing())   
```

Putting get_letter_grade() inside the Student class is useful because each student object can calculate its own grade. This keeps the code organized and prevents us from rewriting the same grade-checking code every time

```python
class Student:
    def __init__(self, name, major, score):
        self.name = name
        self.major = major
        self._score = score

    def display_info(self):
        print("Name:", self.name)
        print("Major:", self.major)
        print("Score:", self._score)

    def get_letter_grade(self):
        if self._score >= 90:
            return "A"
        elif self._score >= 80:
            return "B"
        elif self._score >= 70:
            return "C"
        elif self._score >= 60:
            return "D"
        else:
            return "F"

    def is_passing(self):
        if self._score >= 60:
            return True
        else:
            return False

    def get_score(self):
        return self._score

    def set_score(self, new_score):
        if new_score >= 0 and new_score <= 100:
            self._score = new_score
            print("Score updated to", new_score)
        else:
            print("Invalid score. Score must be between 0 and 100.")


student1 = Student("Alice Johnson", "Computer Science", 92)

print("Original score:", student1.get_score())

student1.set_score(95)

student1.set_score(120) 
print("Final score:", student1.get_score())
```

Controlling changes to an object's data is useful because it prevents invalid values from being stored. For example, a student score should not be less than 0 or greater than 100.

```python
class GraduateStudent(Student):
    def __init__(self, name, major, score, research_area):
        super().__init__(name, major, score)
        self.research_area = research_area

    def display_research(self):
        print("Student:", self.name)
        print("Research Area:", self.research_area)

graduate_student1 = GraduateStudent(
    "Maya Patel",
    "Computer Science",
    94,
    "Machine Learning"
)

graduate_student2 = GraduateStudent(
    "Daniel Lee",
    "Cybersecurity",
    88,
    "Network Security"
)


graduate_student1.display_research()
print("Letter Grade:", graduate_student1.get_letter_grade())
print("Passing:", graduate_student1.is_passing())

graduate_student2.display_research()
print("Letter Grade:", graduate_student2.get_letter_grade())
print("Passing:", graduate_student2.is_passing())
```

GraduateStudent receives the name, major, and score attributes from the Student class. It also inherits methods such as display_info(), get_letter_grade(), is_passing(), get_score(), and set_score(). The GraduateStudent class adds a research area and a method to display it.

```python
class GraduateStudent(Student):
    def __init__(self, name, major, score, research_area):
        super().__init__(name, major, score)
        self.research_area = research_area

    def display_research(self):
        print("Student:", self.name)
        print("Research Area:", self.research_area)

    def display_info(self):
        print("Name:", self.name)
        print("Major:", self.major)
        print("Score:", self._score)
        print("Research Area:", self.research_area)
student1 = Student("Alice Johnson", "Computer Science", 92)
graduate_student1 = GraduateStudent(
    "Maya Patel",
    "Computer Science",
    94,
    "Machine Learning"
)
print("Student:")
student1.display_info()
print("Graduate Student:")
graduate_student1.display_info()
```

Method overriding means that a subclass creates its own version of a method that already exists in the parent class. In this example, GraduateStudent overrides display_info() so that it also displays the student's research area.

```python
students = [
    student1,
    student2,
    graduate_student1,
    graduate_student2
]

for student in students:
    student.display_info()
    print()
```

This example shows polymorphism because the same method call, display_info(), works with different types of objects. Regular Student objects use the Student version of the method, while GraduateStudent objects use the overridden GraduateStudent version.

```python

student1 = Student("Alice Johnson", "Computer Science", 92)
student2 = Student("John Mavi", "Cybersecurity", 70)
student3 = Student("Tanner Baucknight", "Marketing", 50)
graduate_student1 = GraduateStudent(
    "Maya Patel",
    "Computer Science",
    94,
    "Machine Learning"
)
graduate_student2 = GraduateStudent(
    "Daniel Lee",
    "Cybersecurity",
    88,
    "Network Security"
)


students = [
    student1,
    student2,
    student3,
    graduate_student1,
    graduate_student2
]
print("All Students:")
for student in students:
    student.display_info()
    print()

total = 0
for student in students:
    total = total + student.get_score()

average = total / len(students)
print("Average student score:", format(average, ".2f"))
passing_count = 0
for student in students:
    if student.is_passing():
        passing_count = passing_count + 1

print("Number of passing students:", passing_count)
highest_student = students[0]
for student in students:
    if student.get_score() > highest_student.get_score():
        highest_student = student
print("Highest score:", highest_student.name, "with",
      highest_student.get_score())
for student in students:
    if isinstance(student, GraduateStudent):
        print(
            "Graduate Student:",
            student.name,
            "-",
            student.research_area
        )
```

The student data shows that most of the students are passing, although there is a difference between the highest and lowest scores. The graduate students also performed well, and Maya Patel had the highest score in the group.
